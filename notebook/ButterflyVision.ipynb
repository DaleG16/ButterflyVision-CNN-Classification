```python
# To count no of images in in each folder of training dataset,val and test dataset

def count_images(directory):
    class_counts={}
    for class_folder in sorted(directory.iterdir()):
        if class_folder.is_dir():
            # Iterate over each class folder to get count of images
            images=[file for file in class_folder.iterdir() # goes inside each class
                    if file.suffix.lower() in ['.jpg','.jpeg','.png']
                   ]
        class_counts[class_folder.name]=len(images)
    return class_counts

train_counts=count_images(train_dir)
validation_counts=count_images(valid_dir)
test_counts=count_images(test_dir)


    
  
```

```python
def find_corrupted_images(directory):
    corrupted = []

    for class_folder in directory.iterdir():
        if not class_folder.is_dir(): #Ignore if its  not folder(ex:Readme.txt)
            continue

        for image_path in class_folder.iterdir():

            if image_path.suffix.lower() not in [".jpg", ".jpeg", ".png"]: #Ignore if its other extension
                continue

            try:
                with Image.open(image_path) as img:
                    img.verify()

            except Exception:
                corrupted.append(image_path)

    return corrupted


corrupted_train = find_corrupted_images(train_dir)
corrupted_valid = find_corrupted_images(valid_dir)
corrupted_test = find_corrupted_images(test_dir)
print("="*50)
print("Checking whether any files are corrupted or not")
print("="*50)
print("Corrupted training images   :", len(corrupted_train))
print("Corrupted validation images :", len(corrupted_valid))
print("Corrupted testing images    :", len(corrupted_test))
```

```python
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator

train_datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=20,
    width_shift_range=0.1, #image can be shifted left or right by up to 10% of its width
    height_shift_range=0.1,
    shear_range=0.2,
    zoom_range=0.1,
    horizontal_flip=True
)

valid_datagen = ImageDataGenerator(
    rescale=1./255
)

test_datagen = ImageDataGenerator(
    rescale=1./255
)
```

```python
#Creating Generators

#A generator is basically a pipeline that continuously prepares batches of data for the CNN.

train_generator = train_datagen.flow_from_directory(
    train_dir,
    target_size=(224, 224),
    batch_size=32,
    class_mode="categorical",
    shuffle=True #Randomly mix the training images instead of always giving them in folder order
    
)

valid_generator = valid_datagen.flow_from_directory(
    valid_dir,
    target_size=(224, 224),
    batch_size=32,
    class_mode="categorical",
    shuffle=False
)

test_generator = test_datagen.flow_from_directory(
    test_dir,
    target_size=(224, 224),
    batch_size=32,
    class_mode="categorical",
    shuffle=False
)
```

```python
images, labels = next(train_generator) #the generator gives us 32 images and their 32 corresponding labels.

print("Image batch shape :", images.shape)
print("Label batch shape :", labels.shape)
print("Pixel minimum     :", images.min())
print("Pixel maximum     :", images.max())
```

```python
#PHASE-3 CUSTOM CNN

from tensorflow.keras.models import Sequential
from tensorflow.keras import layers
from tensorflow.keras.layers import Conv2D,BatchNormalization,MaxPooling2D,GlobalAveragePooling2D

custom_cnn=Sequential()
# Input
custom_cnn.add(layers.Input(shape=(224, 224, 3)))

# Block 1
custom_cnn.add(Conv2D(32,(3,3),padding='same'))
custom_cnn.add(BatchNormalization())
custom_cnn.add(layers.ReLU())
custom_cnn.add(MaxPooling2D(pool_size=(2,2)))

# Block 2
custom_cnn.add(Conv2D(64,(3,3),padding='same'))
custom_cnn.add(BatchNormalization())
custom_cnn.add(layers.ReLU())
custom_cnn.add(MaxPooling2D(pool_size=(2,2)))

# Block 3

custom_cnn.add(Conv2D(128,(3,3),padding='same'))
custom_cnn.add(BatchNormalization())
custom_cnn.add(layers.ReLU())
custom_cnn.add(MaxPooling2D(pool_size=(2,2)))

# Feature aggregation
custom_cnn.add(GlobalAveragePooling2D())

# Classification head
custom_cnn.add(layers.Dense(128,activation='relu'))
custom_cnn.add(layers.Dropout(0.4))
custom_cnn.add(layers.Dense(100,activation='softmax'))
```

```python
custom_cnn.summary()
```

```python
# Compiling the model
from tensorflow.keras.optimizers import Adam
custom_cnn.compile(
    optimizer="adam",
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)
```

```python
from tensorflow.keras.applications import EfficientNetB0
from tensorflow.keras import layers, models
```

```python
efficientnet_base = EfficientNetB0(
    include_top=False,
    weights="imagenet",
    input_shape=(224, 224, 3)
)

efficientnet_base.trainable = False

print("EfficientNetB0 base created.")
```

```python

efficientnet_model = models.Sequential([
    
    layers.Input(shape=(224, 224, 3)),
    
    efficientnet_base,
    
    layers.GlobalAveragePooling2D(),
    
    layers.Dropout(0.3),
    
    layers.Dense(100, activation="softmax")
])
```

```python
import tensorflow as tf
import numpy as np

def make_gradcam_heatmap(img_array, model, base_model, last_conv_layer_name):
    
    # Model that gives:
    # 1. convolutional feature maps
    # 2. final prediction
    grad_model = tf.keras.models.Model(
        inputs=model.inputs,
        outputs=[
            base_model.get_layer(last_conv_layer_name).output,
            model.output
        ]
    )

    with tf.GradientTape() as tape:
        conv_outputs, predictions = grad_model(img_array)
        
        # Get the predicted class
        predicted_class = tf.argmax(predictions[0])
        
        # Score of predicted class
        class_score = predictions[:, predicted_class]

    # Gradients of predicted class with respect to feature maps
    grads = tape.gradient(class_score, conv_outputs)

    # Average gradients across height and width
    pooled_grads = tf.reduce_mean(grads, axis=(1, 2))

    # Remove batch dimension
    conv_outputs = conv_outputs[0]
    pooled_grads = pooled_grads[0]

    # Weight each feature map by its importance
    heatmap = conv_outputs @ pooled_grads[..., tf.newaxis]
    heatmap = tf.squeeze(heatmap)

    # ReLU: keep only positive influence
    heatmap = tf.maximum(heatmap, 0)

    # Normalize between 0 and 1
    heatmap /= tf.math.reduce_max(heatmap) + 1e-8

    return heatmap.numpy(), predicted_class.numpy()
```

```python
from pathlib import Path
from PIL import Image
import numpy as np
import random

image_files = list(test_dir.rglob("*.jpg"))

sample_test_image = random.choice(image_files)

print("Random image:", sample_test_image)


# Load and resize
img = Image.open(sample_test_image).convert("RGB")
img = img.resize((224, 224))

# Convert to NumPy and create batch
img_array = np.array(img, dtype=np.float32)
img_array = np.expand_dims(img_array, axis=0)

print("Input shape:", img_array.shape)
```

```python
import tensorflow as tf
import numpy as np

# Getting the actual EfficientNet base
base_model = efficientnet_model.get_layer("efficientnetb0")

# Getting the last convolutional layer
last_conv_layer = base_model.get_layer("top_conv")

# Getting the final classification layer
classifier = efficientnet_model.get_layer("dense_2")

# Building Grad-CAM model using the BASE MODEL, not the Sequential wrapper
grad_model = tf.keras.models.Model(
    inputs=base_model.input,
    outputs=[
        last_conv_layer.output,
        base_model.output
    ]
)

print("Grad-CAM model created successfully.")
```

```python
with tf.GradientTape() as tape:

    # Get convolutional features and EfficientNet features
    conv_outputs, features = grad_model(img_array)

    # Reproduce the classification head
    pooled_features = tf.keras.layers.GlobalAveragePooling2D()(features)

    predictions = classifier(pooled_features)

    predicted_class = tf.argmax(predictions[0])

    class_score = predictions[:, predicted_class]

grads = tape.gradient(class_score, conv_outputs)

pooled_grads = tf.reduce_mean(
    grads,
    axis=(1, 2)
)

conv_outputs = conv_outputs[0]
pooled_grads = pooled_grads[0]

heatmap = conv_outputs @ pooled_grads[..., tf.newaxis]

heatmap = tf.squeeze(heatmap)

heatmap = tf.maximum(heatmap, 0)

heatmap /= tf.math.reduce_max(heatmap) + 1e-8

```

```python
import matplotlib.pyplot as plt
import cv2
import numpy as np

# -----------------------------------
# 1. Get prediction
# -----------------------------------

class_names = {
    v: k for k, v in efficient_test_generator.class_indices.items()
}

predictions = efficientnet_model.predict(
    img_array,
    verbose=0
)

predicted_index = np.argmax(predictions[0])
predicted_name = class_names[predicted_index]
confidence = predictions[0][predicted_index] * 100


# -----------------------------------
# 2. Resize Grad-CAM heatmap
# -----------------------------------

heatmap_resized = cv2.resize(
    heatmap.numpy(),
    (224, 224)
)

heatmap_uint8 = np.uint8(
    255 * heatmap_resized
)

heatmap_color = cv2.applyColorMap(
    heatmap_uint8,
    cv2.COLORMAP_JET
)


# -----------------------------------
# 3. Create overlay
# -----------------------------------

original = np.uint8(img_array[0])

overlay = cv2.addWeighted(
    original,
    0.6,
    heatmap_color,
    0.4,
    0
)


# -----------------------------------
# 4. Display
# -----------------------------------

plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(original)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(
    cv2.cvtColor(
        heatmap_color,
        cv2.COLOR_BGR2RGB
    )
)
plt.title("Grad-CAM Heatmap")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(
    cv2.cvtColor(
        overlay,
        cv2.COLOR_BGR2RGB
    )
)

plt.title(
    f"Prediction: {predicted_name}\n"
    f"Confidence: {confidence:.2f}%"
)

plt.axis("off")

plt.tight_layout()
plt.show()




```
